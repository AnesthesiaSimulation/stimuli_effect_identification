```python
import pandas as pd
import numpy as np
from scipy.stats import shapiro
```

```python
data = pd.read_csv('data/database.csv')

perso_data = pd.read_csv("https://api.vitaldb.net/cases", decimal='.')  # personal data
caseid_list = data.caseid.unique()

perso_data = perso_data.query('caseid in @caseid_list')
```

```python
def print_data(df,name):
    print(f"{name}: {df[name].median():.0f} [{df[name].min():.0f}-{df[name].max():.0f}]")

print_data(perso_data,'age')
print_data(perso_data,'weight')
print_data(perso_data,'height')
print_data(perso_data,'asa')
print(f"sex distribution {(perso_data['sex']=='F').sum()} Female, {(perso_data['sex']=='M').sum()} Male")
print(f"Hypertension proportion: {perso_data['preop_htn'].sum()/len(perso_data):.2%}")
print(f"Diabete proportion: {perso_data['preop_dm'].sum()/len(perso_data):.2%}")
print(f"Kidney injury proportion: {(perso_data['preop_cr']>2).sum()/len(perso_data):.2%}")
```

```python
propo_conc_intub = []
propo_conc_surg = []
remif_conc_intub = []
remif_conc_surg = []
for caseid in caseid_list:
    data_case = data.query('caseid == @caseid')
    intubation = data_case['intubation_time'].values[0]
    surgery = data_case['opstart'].values[0]
    propo_conc_intub.append(data_case.query('Time==@intubation')['Ce_prop'].values[0])
    propo_conc_surg.append(data_case.query('Time==@surgery')['Ce_prop'].values[0])
    remif_conc_intub.append(data_case.query('Time==@intubation')['Ce_remi'].values[0])
    remif_conc_surg.append(data_case.query('Time==@surgery')['Ce_remi'].values[0])

print(f"Propofol concentration at intubation: {np.mean(propo_conc_intub):.2f} ({np.std(propo_conc_intub)}) mg/L")
print(f"Propofol concentration at surgery: {np.mean(propo_conc_surg):.2f} ({np.std(propo_conc_surg)}) mg/L")
print(f"Remifentanil concentration at intubation: {np.mean(remif_conc_intub):.2f} ({np.std(remif_conc_intub)}) ng/mL")
print(f"Remifentanil concentration at surgery: {np.mean(remif_conc_surg):.2f} ({np.std(remif_conc_surg)}) ng/mL")
```
