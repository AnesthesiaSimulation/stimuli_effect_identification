```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import python_anesthesia_simulator as pas
```

```python
disturbance = pas.Disturbances(dist_profil='VitalDB')

t = np.linspace(0,90,200)*60
dist = disturbance.compute_dist(t)

dist_tpr = dist[3]
dist_sv = dist[4]
dist_hr = dist[5]

plt.plot(t/60, dist_tpr*10000, label='TPR stimuli (x 10 000)')
plt.plot(t/60, dist_sv, label='SV stimuli')
plt.plot(t/60, dist_hr, label='HR stimuli')
y_lim = plt.gca().get_ylim()
plt.vlines(3 ,y_lim[0], y_lim[1], 'black')
plt.vlines(40,y_lim[0], y_lim[1], 'r')
plt.legend()
plt.xlabel('Time (min)')
plt.ylabel('Stimuli input')
plt.grid()
plt.savefig(f'output/stimuli_show.pdf', dpi=300)
plt.show()
```
